## LinkedIn Post Generator using LangChain

**Objective:** Build an AI Agent with LangChain that generates professional LinkedIn posts based on:
- **Topic** (e.g., "AI in Healthcare", "Remote Work Productivity")
- **Language** (e.g., English, Bengali, Spanish)

**How it works:** A LangChain LLM Chain takes the topic + language, formats them into a
structured prompt template, sends it to an LLM, and returns a polished, ready-to-publish
LinkedIn post (2–4 paragraphs, with a hook, body, and call-to-action/hashtags).

## 3️⃣ Build the LangChain LLM Chain (the "AI Agent")

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate

#The LLM>>
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
)

#The Prompt Templat>>
linkedin_prompt = PromptTemplate(
    input_variables=["topic", "language"],
    template="""You are an expert LinkedIn content strategist and copywriter.

Write a professional, engaging LinkedIn post about the following topic:

Topic: {topic}
Language: {language}

Requirements:
- Write the ENTIRE post in {language}.
- Structure: a strong hook (1-2 lines), a body with 2-4 short paragraphs offering real
  insight/value, and a closing line with a call-to-action or thought-provoking question.
- Tone: professional, confident, conversational — NOT overly salesy or robotic.
- Add line breaks between paragraphs (LinkedIn posts are skimmable, not walls of text).
- End with 3-5 relevant hashtags in {language} or English, whichever fits naturally.
- Do not include any preamble like "Here is your post" — output ONLY the post itself.
"""
)


linkedin_chain = linkedin_prompt | llm

print("LangChain chain ready ✅")
```

## 4️⃣ Wrap It in a Simple Agent Function

```python
def generate_linkedin_post(topic: str, language: str) -> str:
    """
    Generate a LinkedIn post using the LangChain Gemini agent.
    """

    if not topic.strip():
        raise ValueError("Topic cannot be empty.")

    if not language.strip():
        raise ValueError("Language cannot be empty.")

    result = linkedin_chain.invoke({
        "topic": topic,
        "language": language
    })

    # Handle different content formats returned by the chat model
    if isinstance(result.content, str):
        return result.content.strip()

    if isinstance(result.content, list):
        text_parts = []

        for block in result.content:
            if isinstance(block, dict) and "text" in block:
                text_parts.append(block["text"])
            elif isinstance(block, str):
                text_parts.append(block)

        return "\n".join(text_parts).strip()

    return str(result.content).strip()


print("generate_linkedin_post() function ready ✅")
```

## 5️⃣ Demo Run (fixed examples — good to show in your video)

```python
demo_examples = [
    {"topic": "AI in Healthcare", "language": "English"},
    {"topic": "Remote Work Productivity", "language": "Bengali"},
]

for ex in demo_examples:

    print(
        f"=== Topic: {ex['topic']} | "
        f"Language: {ex['language']} ===\n"
    )

    post = generate_linkedin_post(
        ex["topic"],
        ex["language"]
    )

    print(post)

    print("\n" + "-" * 70 + "\n")
```
